```python
from tokenize import ContStr
import pandas as pd

# Đọc dữ liệu từ tệp CSV
df = pd.read_csv('train1.csv')

education_map = {
    1: "work full time", 
    2: "work part time",
}


male_map = {
    1: "male",
    2: "female"
}

# Hàm để tạo chuỗi văn bản từ một hàng dữ liệu

def create_text(row):
    age = "..." if row["age"] == -1 else int(row["age"])
    gender = "..." if row["Gender"] == -1 else male_map.get(int(row["Gender"]))
    education = "... education" if row["OccupStat"] == -1 else education_map.get(int(row["OccupStat"]))

    distance = "..." if row["distance_km"] == -1 else float(row["distance_km"])

    # phương tiện trong hộ
    if row["NbCar"] == -1:
        car = "... car"
    else:
        c = int(row["NbCar"]) 
        car = f"{c} cars" if c > 1 else f"{c} car"

    if row["NbMoto"] == -1:
        motor = "... motorcycle"
    else:
        m = int(row["NbMoto"]) 
        motor = f"{m} motorcycles" if m > 1 else f"{m} motorcycle"

    if row["NbBicy"] == -1:
        bicycle = "... bicycle"
    else:
        b = int(row["NbBicy"]) 
        bicycle = f"{b} bicycles" if b > 1 else f"{b} bicycle"
        
    if row["NbChild"] == -1:
        child = "... child"
    else:
        c = int(row["NbChild"]) 
        child = f"{c} children" if c > 1 else f"no children"

    timecar = "..." if row["TimeCar"] == -1 else int(row["TimeCar"])
    timept = "..." if row["TimePT"] == -1 else int(row["TimePT"])

    costcar = "..." if row["CostCarCHF"] == -1 else float(row["CostCarCHF"])
    costpt = "..." if row["MarginalCostPT"] == -1 else float(row["MarginalCostPT"])

        
    return (
        f"The trip distance is {distance} km. There are three travel modes for the traveler to choose from: "
        f"1. Public transport: travel time is {timept} minutes, and the travel cost is {costpt} CHF. "
        f"2. Driving: travel time is {timecar} minutes, and the travel cost is {costcar} CHF. "
        f"3. Soft modes are free of charge. "
        f"Traveler is {age} year old {gender} who {education} and has {child}. "
        f"The household owns {car}, {motor}, {bicycle}."
    )

#Test
print(create_text(df.iloc[0]))
# Tạo cột văn bản mới
df['INFOR'] = df.apply(create_text, axis=1)
df['CHOICE'] = df['Choice'].astype(int)

# # Lưu kết quả vào tệp CSV mới
df[['ID', 'INFOR', 'CHOICE']].to_csv('train1.csv', index=False)

print("Dữ liệu đã được chuyển đổi và lưu vào 'text_data.csv'")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Đọc dữ liệu từ file
df = pd.read_csv("textdata.csv")  # Đảm bảo file nằm cùng thư mục với script

# Chia dữ liệu theo tỉ lệ 70:30 (train:test)
train_df, test_df = train_test_split(df, test_size=0.3, stratify=df['CHOICE'], random_state=42)

# Lưu ra file
train_df.to_csv("train.csv", index=False)
test_df.to_csv("test.csv", index=False)

# In thông tin về kích thước các tập dữ liệu
print(f"Tổng số mẫu: {len(df)}")
print(f"Train set: {len(train_df)} mẫu ({len(train_df)/len(df)*100:.1f}%)")
print(f"Test set: {len(test_df)} mẫu ({len(test_df)/len(df)*100:.1f}%)")
```

```python
from tokenize import ContStr
import pandas as pd

df=pd.read_csv("train.csv")

df["Gender"]=df["Gender_man"]
df.loc[df["Gender_woman"] == 1, "Gender"] = 2
df.loc[df["Gender_unreported"] == 1, "Gender"] = -1
df.drop(columns=["Gender_man", "Gender_woman", "Gender_unreported"], inplace=True)

df["OccupStat"]=df["OccupStat_fulltime"]
df.loc[df["OccupStat_notfulltime"] == 1, "OccupStat"] = 2
df.drop(columns=["OccupStat_fulltime", "OccupStat_notfulltime"], inplace=True)

df.to_csv("train.csv", index=False)
```
